```python
%sql
-- ============================================================
-- SENTINEL COMMERCE
-- Notebook: 00_platform_setup
-- Purpose : Create the foundational Lakehouse structure
-- ============================================================

CREATE CATALOG IF NOT EXISTS sentinel_dev;

USE CATALOG sentinel_dev;

-- Raw ingestion layer
CREATE SCHEMA IF NOT EXISTS bronze;

-- Validated / transformed layer
CREATE SCHEMA IF NOT EXISTS silver;

-- Business / dimensional layer
CREATE SCHEMA IF NOT EXISTS gold;

-- Data quality, pipeline metrics, observability
CREATE SCHEMA IF NOT EXISTS monitoring;

-- Landing area for incoming source files
CREATE SCHEMA IF NOT EXISTS landing;
```
