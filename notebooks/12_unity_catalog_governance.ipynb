# Sentinel Lakehouse — Unity Catalog Governance

## Governance Model

Sentinel uses Unity Catalog as the governance boundary for its
lakehouse architecture.

### Catalog

sentinel_dev

### Data Domains

- landing — source-file ingestion
- bronze — raw ingestion layer
- silver — validated and conformed operational data
- gold — curated analytical data
- monitoring — pipeline health and operational metrics

## Intended Production Access Model

### Data Engineers
- USE CATALOG
- USE SCHEMA
- SELECT
- MODIFY
- CREATE TABLE
- CREATE VOLUME where ingestion requires it

Engineering workloads own ingestion, transformation, data-quality,
CDC and Gold processing.

### Analysts
- USE CATALOG on sentinel_dev
- USE SCHEMA on sentinel_dev.gold
- SELECT on Gold tables

Analysts should not receive direct write access to Bronze, Silver,
or monitoring datasets.

### Pipeline / Service Principal
Production automation should execute using a dedicated service
principal rather than an individual user identity.

It requires only the privileges necessary to read upstream datasets
and write its managed downstream datasets.

### Monitoring / Operations
Operational users receive read access to the monitoring schema
without requiring write access to business datasets.

## Principle of Least Privilege

Access is separated by responsibility:

Engineering → processing layers
Analytics   → curated Gold
Operations  → monitoring

The portfolio workspace is a single-user development environment,
so the production RBAC model is documented rather than simulated
using artificial user accounts.
